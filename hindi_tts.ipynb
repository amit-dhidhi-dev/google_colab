```python
# सेल 1: VITS Model for Hindi
!pip install torch torchaudio
!pip install transformers
!pip install soundfile

print("Installing VITS...")

# सेल 2: VITS Hindi Model
from transformers import VitsModel, AutoTokenizer
import torch
import scipy.io.wavfile
import IPython.display as ipd

# Load Hindi VITS model
model_id = "facebook/mms-tts-hin"  # Facebook's Hindi TTS model

try:
    model = VitsModel.from_pretrained(model_id)
    tokenizer = AutoTokenizer.from_pretrained(model_id)

    # Hindi text
    text = "नमस्ते, यह हिंदी में टेक्स्ट टू स्पीच है"

    # Tokenize
    inputs = tokenizer(text, return_tensors="pt")

    # Generate speech
    with torch.no_grad():
        output = model(**inputs).waveform

    # Save and play
    scipy.io.wavfile.write("hindi_vits.wav", rate=model.config.sampling_rate, data=output.float().numpy().T)

    print("Hindi VITS audio generated successfully!")
    ipd.Audio("hindi_vits.wav")

except Exception as e:
    print(f"Error loading Hindi VITS: {e}")
    print("Trying alternative model...")

    # Try English model
    model_id = "facebook/mms-tts-eng"
    model = VitsModel.from_pretrained(model_id)
    tokenizer = AutoTokenizer.from_pretrained(model_id)

    text = "Hello, this is English text to speech"
    inputs = tokenizer(text, return_tensors="pt")

    with torch.no_grad():
        output = model(**inputs).waveform

    scipy.io.wavfile.write("english_vits.wav", rate=model.config.sampling_rate, data=output.float().numpy().T)
    ipd.Audio("english_vits.wav")
```

```python
# सेल 1: पाइथन TTS लाइब्रेरी
!pip install pyttsx3
!apt-get install -y espeak-ng

import pyttsx3
import IPython.display as ipd
import numpy as np
import wave

print("pyttsx3 installed!")

# सेल 2: Offline Hindi TTS
# Initialize TTS engine
engine = pyttsx3.init()

# Try to set Hindi voice
voices = engine.getProperty('voices')

# Print available voices
print("Available voices:")
for i, voice in enumerate(voices):
    print(f"{i}: {voice.name} - {voice.languages}")

# Try to find Hindi voice
hindi_voice_found = False
for voice in voices:
    if 'hindi' in voice.name.lower() or 'hi' in str(voice.languages).lower():
        engine.setProperty('voice', voice.id)
        hindi_voice_found = True
        print(f"Hindi voice found: {voice.name}")
        break

if not hindi_voice_found:
    print("Hindi voice not found, using default voice")
    # Use default voice

# Set properties
engine.setProperty('rate', 150)  # Speed
engine.setProperty('volume', 0.9)  # Volume

# Hindi text
hindi_text = "नमस्ते, यह हिंदी टेक्स्ट टू स्पीच का परीक्षण है"

# Save to file
output_file = "pyttsx_hindi.wav"
engine.save_to_file(hindi_text, output_file)
engine.runAndWait()

# Play the audio
print("Playing generated audio...")
ipd.Audio(output_file)
```

```python
# Complete Working Solution - Facebook MMS Hindi TTS
!pip install -q torch torchaudio transformers soundfile
!apt-get install -y ffmpeg

import torch
from transformers import VitsModel, AutoTokenizer
import numpy as np
from scipy.io.wavfile import write
import IPython.display as ipd
import warnings
warnings.filterwarnings('ignore')

print("Hindi TTS System Ready!")

def hindi_tts(text, output_file="hindi_output.wav"):
    """
    Convert Hindi text to speech using Facebook MMS
    """
    try:
        # Load Hindi model
        model = VitsModel.from_pretrained("facebook/mms-tts-hin")
        tokenizer = AutoTokenizer.from_pretrained("facebook/mms-tts-hin")

        # Tokenize text
        inputs = tokenizer(text, return_tensors="pt")

        # Generate speech
        with torch.no_grad():
            waveform = model(**inputs).waveform

        # Convert to numpy and save
        audio_np = waveform.float().numpy().T
        sampling_rate = model.config.sampling_rate

        # Save to file
        write(output_file, sampling_rate, audio_np)

        return output_file, sampling_rate

    except Exception as e:
        print(f"Hindi model error: {e}")

        # Fallback to English
        print("Using English as fallback...")
        model = VitsModel.from_pretrained("facebook/mms-tts-eng")
        tokenizer = AutoTokenizer.from_pretrained("facebook/mms-tts-eng")

        inputs = tokenizer("Hello, this is English text", return_tensors="pt")

        with torch.no_grad():
            waveform = model(**inputs).waveform

        audio_np = waveform.float().numpy().T
        sampling_rate = model.config.sampling_rate

        write(output_file, sampling_rate, audio_np)

        return output_file, sampling_rate

# Test Hindi TTS
test_texts = [
    "नमस्ते, आप कैसे हैं?",
    "आज मौसम बहुत सुहावना है।",
    "धन्यवाद और शुभकामनाएँ।"
]

for i, text in enumerate(test_texts):
    print(f"\n{'='*50}")
    print(f"Text {i+1}: {text}")
    print(f"{'='*50}")

    output_file, sr = hindi_tts(text, f"hindi_{i+1}.wav")

    print(f"Generated: {output_file}")
    display(ipd.Audio(output_file))
    print("\n")

print("All audio files generated successfully!")
```
